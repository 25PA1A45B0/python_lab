```python
#OPERATIONS
t = (1, 2, 3, 55.6, 'hello')
t1 = (4, 5, 6)
print(t[1]) #indexing
print(t[1] + t1[1]) #concatenation
print(t1 * 2) #repetition
print(t + t1) #concatenation
if 'hello' in t:  #membership
  print("YES")
else:
  print("NO")
```

```python
#METHODS
t2 = (1,2,3,4,5,6,7,2,67,4)
print(t2.count(2)) #count
print(t2.index(5)) #index

```

```python
#FUNCTIONS
print(len(t2)) #length
print(max(t2)) #maximum
print(min(t2)) #minimum
print(sum(t2)) #sum
print(sorted(t2)) #sorted
print(sorted(t2, reverse=True)) #sorted in reverse
```

SETS IN PYTHON


```python
#operations
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(A|B)  #union
print(A&B)  #intersection
print(A-B)  #difference
print(A^B)  #symmetric diifference
```

```python
#methods
A = {1, 2, 3, 4, 5}
A.add(6)
print(A)
A.remove(6)  #removes 6
print(A)
A.discard(7)   #7 is not there, but no error will be displayed
print(A)
A.pop()  #removes first element
print(A)
A.clear()  #clears all elements
print(A)
A.update({1, 2, 3, 4, 5})  #updates the set
print(A)
print(A.issubset({6, 7, 8}))
```
