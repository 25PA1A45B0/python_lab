```python
#examples on string methods
a = 'hello world'
b = a.split()
print(b)
print("/".join(b))
x =" ai & ds "
print(x.strip())
print(a.upper())
print(a.lower())
y = a.replace("hello","bye")
print(y)
print(a.find("h"))
print(a.count("o"))
```

```python
#program that count the no of vowels and consonants in a string
a = input()
b = a.lower()
vowels = "aeiou"
count= 0
fount= 0
for i in b:
  if i in vowels:
    count = count+1
  else:
    fount+=1
print(count)
print(fount)
```
