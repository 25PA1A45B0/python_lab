```python
#Lists in python
L1 = [11,22,33]
L2 = [33,44,55]
print(L1)
```

```python
#concatination
print(L1+L2)
```

```python
#Repetition
print(L1*3)
```

```python
#Membership
print(2 in L1)
```

```python
#slicing
A = [1,2,3,4,5,"hello"]
print(A[2:7])
print(A[::-1]) #prints elements in reverse order
```

```python
print(2 not in L1)
```

```python
#iteration
for i in L1:
  print(i)
```
