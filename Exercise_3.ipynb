For loop in python

```python
for i in range(1,11):
  print("python")
```

```python
#printing 1 to n odd numbers
n = int(input());
for i in range(1,n+1):
  print(i)
```

```python
#print multiplication table of given number
n = int(input());
for i in range(1,11):
  print(n,"*",i ,"=",n*i)
```

```python
#check whether the number is prime or not
n = int(input());
for i in range(2,n):
  if(n%i == 0):
    print("It is not a prime")
    break
else:
  print("It is a prime")
```

```python
#print prime numbers between 2 to no's
a = int(input());
b = int(input());
for i in range(a,b+1):
  for j in range(2,i):
   if(i%j == 0):
    break
  else:
   print(i)
```

```python
#sum of the digits of a given number
n = int(input());
sum = 0
while(n>0):
  digit = n%10
  sum = sum +digit
  n = n//10
print(sum)
```

```python
#printing 1 to 10 numbers 10 times
for i in range(1,11):
  for j in range(1,11):
    print(j,end = " ")
  print("\n")
```
